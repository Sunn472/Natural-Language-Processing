# REGEX

| Function / Symbol | Meaning                       | Example                      | Result / Use                 |
| ----------------- | ----------------------------- | ---------------------------- | ---------------------------- |
| `re.findall()`    | Find **all** matches          | `re.findall(r"\d", data)`    | Find all digits              |
| `re.search()`     | Find **first** match anywhere | `re.search(r"\d+", data)`    | Find first number            |
| `re.match()`      | Match at **beginning**        | `re.match(r"Hello", data)`   | Checks starting text         |
| `re.fullmatch()`  | Match **entire string**       | `re.fullmatch(r"\d+", data)` | Entire string must be digits |
| `re.sub()`        | Replace matches               | `re.sub(r"\d+", "", data)`   | Remove digits                |
| `re.split()`      | Split using pattern           | `re.split(r"\s+", data)`     | Split by spaces              |


- **Regex Symbols**
| Regex | Meaning            | Example   | Meaning in Example         |                    |                |
| ----- | ------------------ | --------- | -------------------------- | ------------------ | -------------- |
| `\d`  | Digit              | `\d`      | `0–9`                      |                    |                |
| `\D`  | Non-digit          | `\D`      | Anything except `0–9`      |                    |                |
| `\w`  | Word character     | `\w`      | `A-Z, a-z, 0-9, _`         |                    |                |
| `\W`  | Non-word character | `\W`      | `@, #, !, etc.`            |                    |                |
| `\s`  | Whitespace         | `\s`      | Space, tab, newline        |                    |                |
| `\S`  | Non-whitespace     | `\S`      | Anything except whitespace |                    |                |
| `.`   | Any character      | `.`       | `a`, `1`, `@`, etc.        |                    |                |
| `^`   | Beginning          | `^Hello`  | Starts with `Hello`        |                    |                |
| `$`   | End                | `World$`  | Ends with `World`          |                    |                |
| `\b`  | Word boundary      | `\bcat\b` | Exact word `cat`           |                    |                |
| `[]`  | Character set      | `[aeiou]` | Any vowel                  |                    |                |
| `[^]` | NOT in set         | `[^0-9]`  | Anything except digit      |                    |                |
| `()`  | Group              | `(cat     | dog)`                      | Group alternatives |                |
| `     | `                  | OR        | `cat                       | dog`               | `cat` OR `dog` |


- **Quantifiers**
| Symbol  | Meaning       | Example   | Meaning             |
| ------- | ------------- | --------- | ------------------- |
| `+`     | **1 or more** | `\d+`     | `123`               |
| `*`     | **0 or more** | `a*`      | ``, `a`, `aaa`      |
| `?`     | **0 or 1**    | `colou?r` | `color` / `colour`  |
| `{n}`   | Exactly `n`   | `\d{4}`   | `2026`              |
| `{n,m}` | `n` to `m`    | `\d{2,4}` | `12`, `123`, `1234` |
| `{n,}`  | `n` or more   | `\d{3,}`  | `123`, `12345`, ... |

| Task                   | Code                                        |
| ---------------------- | ------------------------------------------- |
| Find digits            | `re.findall(r"\d+", data)`                  |
| Find words             | `re.findall(r"\w+", data)`                  |
| Find spaces            | `re.findall(r"\s+", data)`                  |
| Remove digits          | `re.sub(r"\d+", "", data)`                  |
| Remove punctuation     | `re.sub(r"[^\w\s]", "", data)`              |
| Split words            | `re.split(r"\s+", data)`                    |
| Find uppercase letters | `re.findall(r"[A-Z]", data)`                |
| Find lowercase letters | `re.findall(r"[a-z]", data)`                |
| Find email             | `re.findall(r"[\w.-]+@[\w.-]+\.\w+", data)` |


**⭐ Remember these 6 first**

- re.findall()    → Find ALL
- re.search()     → Find FIRST
- re.match()      → Find at START
- re.fullmatch()  → Match WHOLE string
- re.sub()        → REPLACE
- re.split()      → SPLIT

**And:**
- \d  → digit
- \w  → word
- \s  → space
- +   → one or more
- *   → zero or more
- ?   → optional
- ^   → start
- $   → end
- []  → character set
- |   → OR

- **inflect is a Python library used to convert and generate different forms of English words, especially plural/singular forms, numbers, ordinals, and articles**

```python
# import the library
import inflect
q = inflect.engine()

# convert number into text
def convert_num(text):
  # split strings into list of texts
  temp_string = text.split()
  # intialise empty list
  new_str = []

  for word in temp_string:
    # if text is a digit, convert the digit
    # to numbers and append into the new_str list
    if word.isdigit():
      temp = q.number_to_words(word)
      new_str.append(temp)

    # else the texts as it is
    else:
      new_str.append(word)

  temp_str = ' '.join(new_str)
  return temp_str


input_str = 'You bought 6 candies from shop, and 4 candies are in home.'
convert_num(input_str)
```

```python
for word in input_str:
  print(word)
```

```python
for word in input_str.split():
  print(word)
```

```python
cleaned_sentence = []
for word in input_str.split():
  if word.isdigit():
    temp = q.number_to_words(word)
    cleaned_sentence.append(temp)
  else:
    cleaned_sentence.append(word)
cleaned_sentence
```

```python
new = " ".join(cleaned_sentence)
new
```

```python
def remove_punctuation(text):
  # creating translation
  translator = str.maketrans("","",string.punctuation)#"","",Nothing to change
  print(translator)
  return text.translate(translator)

#text.translate(table,/)
#replace each character in the string using the gie=ven translation table
```
