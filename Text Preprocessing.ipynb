## **Tokenization**

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

#sample text

text = "I love python programming"

tokens = word_tokenize(text)

print(tokens)
```

```python
from nltk import sent_tokenize

#sample text

text = """
My name is XYZ, i am a big fan of python programming.
I wanted to learn NLP, and i have come to the right place.
I hope, i will be making a lot of applications in no time.
"""

tokens = sent_tokenize(text)

print(tokens)
```

```python
for i in tokens:
  new_tokens = word_tokenize(i)
  print(new_tokens)
```

# **Remove Punctuation**

```python
import string

#sample text
text = """
I have a news, and it is great. I am going to share it, but only with people
who have helped me. And especially those, who have helped me stay at the top.
"""

translator = str.maketrans('','', string.punctuation)
processed_text = text.translate(translator)

print(processed_text)
```

# **Stemming**

```python
from nltk.stem import PorterStemmer

#sample words
words = ['runner', 'burnt', 'happily', 'easily', 'lovely', 'flying', 'leaf']

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

# **Lemmatization**

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = "i am running my best time"

doc = nlp(text)

lemmatized_words = [token.lemma_ for token in doc]
print(lemmatized_words)
```

```python
for token in doc:
  print(f"{token.text:<15}{token.pos_:<10}{token.lemma_}")
```

# **N-Grams**

```python
from nltk.util import ngrams
```

```python
sample_text = "My name is XYZ, i have a 10 sec record in running 100 metres"

#tokenize
tokens = word_tokenize(sample_text)


#generate_n=10grams

tengrams = list(ngrams(tokens, 10))
print(tengrams)
```

```python
len(tokens)
```

```python
len(tengrams)
```

```python
sample_text = "My name is not XYZ, i haven't set a 10 sec record in running 100 metres"
sample_text = sample_text.lower()

processed = sample_text.translate(translator)
tokenize = word_tokenize(processed)

sample_text = [word for word in tokenize if word.lower() not in stop_words]

bigrams = list(ngrams(sample_text, 2))

print(sample_text)
print(bigrams)
len(bigrams)
```

# **POS Tagging**
