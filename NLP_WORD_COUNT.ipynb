# **Regex - Handle Punctuation and Count Words**

```python
import re
def count_words(text):
  words = re.findall(r'\b\w+\b',text)
  return len(words)

text = "Manual sprayers, battery-operated power sprayers, engine-operated power sprayers, tractor-mounted boom sprayers, and agricultural drones"
print("Word Count:",count_words(text))
```

# **Spacy**

```python
import spacy
nlp = spacy.load("en_core_web_sm")

def count_words(text):
  doc = nlp(text)
  words = [token.text for token in doc if not token.is_punct]
  return len(words)

text = "Yes, the RBI Grade B exam is widely considered one of the toughest competitive exams in India due to an extremely low success rate"
print("word_count:",count_words(text))
```

# **Counter**

```python
from collections import Counter

def word_frequency(text):
  words = text.lower().split()
  return Counter(words)

text = "NLP is fun. NLP is powerful. NLP is the future."
word_freq = word_frequency(text)
word_count = sum(word_freq.values())
print("Word Frequency:", word_freq)
print("Total Word Count:", word_count)
```

# **NLP Task 1 - Short vs Long sentences**

```python
from nltk.tokenize import sent_tokenize

def categorize_sentence(text):
  sentences = sent_tokenize(text)
  categorized = {"short":[],"long":[]}

  for sent in sentences:
    word_count = len(word_tokenize(sent))
    if word_count < 5:
      categorized["short"].append(sent)
    else:
      categorized["long"].append(sent)
  return categorized


text = "Hello! NLP is fun. Let's learn NLP together. It helps in many fields."
categories = categorize_sentence(text)
print("Short Sentences:", categories["short"])
print("Long Sentences:", categories["long"])
```

# **NLP Task 2 - Rule based Sentiment Analysis**

```python
import time

positive_words = {"happy", "good", "great", "excellent", "love", "amazing"}
negative_words = {"sad", "bad", "terrible", "hate", "awful", "worst"}

def sentiment_analysis(text):
  start = time.time()
  words = text.lower().split()
  pos_count = sum(1 for word in words if word in positive_words)
  neg_count = sum(1 for word in words if word in negative_words)

  sentiment = "Positive Sentiment" if pos_count > neg_count else "Negative Sentiment"

  end = time.time()

  execution_time = end - start

  return sentiment, execution_time

text = """
I'm happy you're the President, and thank you for bringing me home.
I have never been so proud to be an American citizen. Thank you, Mr. President.
"""
sentiment, exec_time = sentiment_analysis(text)
print("Sentiment:", sentiment)
print("Execution Time: {:.6f} seconds".format(exec_time))
```
